```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
```

```python
np.random.seed(42)

num_users = 500

users = pd.DataFrame({
    "user_id": range(1, num_users + 1),
    "age": np.random.randint(18, 60, num_users),
    "gender": np.random.choice(["Male", "Female"], num_users),
    "location": np.random.choice(
        ["India", "USA", "UK", "Germany", "Canada"], num_users
    )
})

users
```

```python
users.to_csv("users.csv", index=False)
```

```python
num_products = 300

products = pd.DataFrame({
    "product_id": range(1, num_products + 1),
    "category": np.random.choice(
        ["Electronics", "Fashion", "Home", "Books", "Beauty"], num_products
    ),
    "price": np.random.randint(200, 5000, num_products),
    "profit_margin": np.round(np.random.uniform(0.05, 0.4, num_products), 2),
    "inventory_level": np.random.randint(10, 500, num_products)
})

products
```

```python
products.to_csv("products.csv", index=False)
```

```python
num_interactions = 20000

interaction_types = ["click", "add_to_cart", "purchase"]

interactions = pd.DataFrame({
    "user_id": np.random.choice(users["user_id"], num_interactions),
    "product_id": np.random.choice(products["product_id"], num_interactions),
    "event_type": np.random.choice(
        interaction_types,
        num_interactions,
        p=[0.6, 0.25, 0.15]
    ),
    "timestamp": [
        datetime.now() - timedelta(days=np.random.randint(0, 180))
        for _ in range(num_interactions)
    ]
})

interactions
```

```python
interactions.to_csv("interactions.csv", index=False)
```
